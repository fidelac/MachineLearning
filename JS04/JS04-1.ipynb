```python
	# import libraries for visualization
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# visualize data with pairplot
sns.pairplot(data, x_vars=['Time on App', 'Time on Website', 'Length of Membership'],
             y_vars='Yearly Amount Spent', size=4, aspect=1, kind='scatter')
plt.show()
```

```python
# Visualize correlation using only numeric columns
sns.heatmap(data.select_dtypes(include=['number']).corr(), cmap="YlGnBu", annot=True)
plt.show()
```

```python
# Create independent variable X and Y, example selection from prior correlation analysis
X = data['Length of Membership']
y = data['Yearly Amount Spent']
```

```python
# Split training and test data with a 7:3 proportion
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.7, test_size=0.3, random_state=100)
```
