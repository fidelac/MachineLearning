### Kegunaan Pustaka

1. **PyPREP**  
   PyPREP digunakan untuk melakukan preprocessing dan persiapan data sebelum data digunakan dalam proses pembelajaran mesin.

2. **SciPy**  
   SciPy merupakan pustaka Python yang digunakan untuk komputasi ilmiah, seperti perhitungan matematika, statistik, optimasi, dan pengolahan data.

3. **tongkat**  
   Tongkat merupakan pustaka pendukung yang digunakan dalam proses pengolahan atau analisis data pada praktikum pembelajaran mesin.

4. **EKG py**  
   EKG py digunakan untuk mengolah dan menganalisis sinyal EKG (elektrokardiogram), terutama data yang berkaitan dengan aktivitas listrik jantung.
